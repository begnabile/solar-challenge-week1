```python
cols_to_check = ['GHI', 'DNI', 'DHI', 'ModA', 'ModB', 'WS', 'WSgust']
for col in cols_to_check:
    df[f'{col}_z'] = zscore(df[col].fillna(df[col].median()))

outliers = df[(df[[f'{col}_z' for col in cols_to_check]].abs() > 3).any(axis=1)]
```

```python
df_cleaned = df.copy()
df_cleaned.to_csv('../data/benin_clean.csv', index=False)
```

```python
df['Timestamp'] = pd.to_datetime(df['Timestamp'])  # if needed

df.set_index('Timestamp')[['GHI', 'DNI', 'DHI', 'Tamb']].plot(figsize=(14, 6))
plt.title("Time Series of Irradiance and Temperature")
plt.show()
```

```python
df.groupby('Cleaning')[['ModA', 'ModB']].mean().plot(kind='bar')
plt.title("ModA & ModB Average Pre/Post Cleaning")
```
